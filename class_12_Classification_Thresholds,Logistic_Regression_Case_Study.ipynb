#**PART 1 — Recap: Probabilities vs Class Labels**

**What Logistic Regression Actually Predicts**

- One of the biggest misconceptions is:

  - Logistic Regression predicts classes.

This is incorrect.

**Logistic Regression predicts:**

- Probability of belonging to Class 1

**Mathematically:**

$$P(Y=1|X)$$

**For example:**
| Customer | Predicted Probability |
| -------- | --------------------- |
| A        | 0.92                  |
| B        | 0.68                  |
| C        | 0.42                  |
| D        | 0.12                  |


**The model itself does NOT decide:**
- Approve
- Reject
- Fraud
- Not Fraud
- Spam
- Not Spam

It only produces probabilities.


**From Probability to Decision**

- Suppose:

$$P(Default)=0.82$$

**The business still needs:**

- Default?
  - YES / NO

We need a decision boundary.

**This boundary is called:**

- Classification Threshold


**Visualization**

```python
print("""
Probability Scale

0 ------------------------ 1

      Threshold = 0.5

0.2 → Class 0
0.3 → Class 0
0.7 → Class 1
0.9 → Class 1
""")
```

**Key Insight**

- The model remains identical.

- Only the threshold changes.

**Same Model**
- Different Decisions


**This is extremely important.**

- Many beginners think:

  - Better performance = Better model

**Often:**

- Better performance = Better threshold

**Example**

- Model outputs:


```python
[0.92, 0.81, 0.67, 0.43, 0.28]
```

**Threshold = 0.5**

```python
[1,1,1,0,0]
```

**Threshold = 0.7**

```python
[1,1,0,0,0]
```

**Same model.**

- Different behavior.

#**PART 2 — Classification Thresholds & Tuning**

**Default Threshold = 0.5**

- Most ML libraries use:

```python
threshold = 0.5
```

**Decision rule:**

```python
probability = 0.73

if probability >= 0.5:
    prediction = 1
else:
    prediction = 0

print(prediction)
```

**Why 0.5?**

- It assumes:

  - Balanced classes
  - Equal error costs
  - Symmetric business objectives

Rarely true in practice.


**Real World Example**

- Suppose:

  - Fraud Rate = 1%

**Then:**

- 99% transactions = legitimate
- 1% transactions = fraud

Using 0.5 may miss most fraud cases.


**Why Threshold Tuning Is Needed**

- Different businesses have different goals.

**Business Examples**
| Scenario            | Preferred Threshold |
| ------------------- | ------------------- |
| Fraud Detection     | Lower               |
| Medical Diagnosis   | Lower               |
| Spam Filtering      | Higher              |
| Credit Approval     | Higher              |
| Terrorist Detection | Extremely Low       |
| Marketing Campaigns | Moderate            |



**Intuition**

- Lower Threshold:

  - Catch More Positives

- Higher Threshold:

  - Require Strong Evidence

**Threshold Tuning Example**
- Predictions:

```python
[0.90,0.70,0.60,0.45,0.30]
```

**Threshold = 0.5**

```python
[1,1,1,0,0]
```

**Threshold = 0.3**

```python
[1,1,1,1,1]
```

**Threshold = 0.8**

```python
[1,0,0,0,0]
```

**Python Example**

```python
import numpy as np

probs = np.array([0.90,0.70,0.60,0.45,0.30])

threshold = 0.5

predictions = (probs >= threshold).astype(int)

print(predictions)
```

**Effect of Threshold**
| Threshold Change | Precision | Recall |
| ---------------- | --------- | ------ |
| Lower Threshold  | ↓         | ↑      |
| Higher Threshold | ↑         | ↓      |


#**PART 3 — Precision–Recall Tradeoff**

**Confusion Matrix Refresher**
|          | Predicted 1 | Predicted 0 |
| -------- | ----------- | ----------- |
| Actual 1 | TP          | FN          |
| Actual 0 | FP          | TN          |


**Precision**

- Measures:

  - How many predicted positives are correct?

**Formula:**

$$Precision=\frac{TP}
{TP+FP}$$

	​
**Recall**

- Measures:

  - How many actual positives were found?

**Formula:**

- $$Recall=\frac{TP}{TP+FN}

	​
**Example**

- Suppose:

  - TP = 80
  - FP = 20
  - FN = 40

- Precision:

- 80/(80+20)=0.80

**Recall:**

- 80/(80+40)=0.67


**F1 Score**

- Balances Precision and Recall.

$$F1=2×\frac{Precision×Recall}
{Precision+Recall}
$$
	​


**Business Interpretation**

**Medical Diagnosis**

- Missing disease:

  - False Negative

Very costly.

**Need:**

- High Recall

Spam Detection

- False alarm:
  - False Positive

Annoying.

**Need:**
- High Precision


**Why Accuracy Fails**
- Dataset:  
  - 1000 samples
  - 10 frauds
  - 990 normal


**Predict everything as normal:**

- Accuracy:

  - 990 / 1000 = 99%

Looks amazing.

  - Actually useless.

#**PART 4 — ROC Curve From Absolute Scratch**

**Definitions**

- True Positive Rate:

$$TPR=\frac{TP}
{TP+FN}
$$
	​


**Also called:**

- Recall

 - False Positive Rate:

$$FPR=\frac{FP}
{FP+TN}$$
	​


What ROC Plots
| Axis   | Metric |
| ------ | ------ |
| X-axis | FPR    |
| Y-axis | TPR    |


**ROC Visualization**

```python
print("""
TPR ↑

1.0 |                *
    |
0.8 |            *
    |
0.6 |        *
    |
0.4 |     *
    |
0.2 |  *
    |
0.0 +--------------------→ FPR
     0      0.5      1
""")
```

**ROC Curve = Threshold Sweep**

- As threshold changes:

  - TPR changes
  - FPR changes

Each threshold generates a point.

Connecting points gives ROC.


**Manual Example**
| Sample | Actual | Probability |
| ------ | ------ | ----------- |
| A      | 1      | 0.95        |
| B      | 0      | 0.85        |
| C      | 1      | 0.60        |
| D      | 0      | 0.40        |
| E      | 1      | 0.30        |

Sort descending.

Move threshold gradually.

Compute TPR and FPR each step.

Plot resulting points.

#**PART 5 — AUC (Area Under Curve)**

**What Is AUC?**

- AUC measures:

  - Probability that a random positive sample receives a higher score than a random negative sample.

**Intuition**

- Suppose:

  - Positive:

   - 0.92

- Negative:

  - 0.41

Positive ranked higher.

Good.

- AUC counts such pairwise wins.


**AUC Interpretation**
| AUC  | Meaning           |
| ---- | ----------------- |
| 1.0  | Perfect           |
| 0.9  | Excellent         |
| 0.8  | Good              |
| 0.7  | Acceptable        |
| 0.5  | Random Guess      |
| <0.5 | Worse Than Random |


#**PART 6 — ROC vs Precision–Recall Curve**

**Why PR Curve Exists**

- ROC includes:

  - True Negatives

- In highly imbalanced datasets:

  - TN dominates

ROC may appear overly optimistic.

**Comparison**
| Aspect            | ROC       | PR Curve  |
| ----------------- | --------- | --------- |
| Uses TN           | Yes       | No        |
| Balanced Data     | Good      | Good      |
| Rare Events       | Weak      | Strong    |
| Fraud Detection   | Not Ideal | Excellent |
| Medical Diagnosis | Not Ideal | Excellent |

#**PART 8 — Threshold Selection Strategies**

**Metric-Based Selection**

- Choose threshold maximizing:

  - F1
  - Recall
  - Precision
  - PR-AUC

depending on objective.

**Example**

```python
from sklearn.metrics import f1_score
```

**Try thresholds:**

```python
0.1
0.2
0.3
...
0.9
```

Choose best.


**Cost-Based Selection**

- Industry favorite.

**Expected Cost:**

$$Cost=(FP×C_
{FP})+(FN×C_
{FN})$$

Choose threshold minimizing cost.


**Example**

- False Positive Cost:

  - $10

- False Negative Cost:

  - $500

Clearly prioritize recall.


**Business Rule-Based**

- Sometimes:

  - Metrics are secondary.

Business constraints dominate.

**Examples:**
- Regulatory compliance
- Legal requirements
- Risk appetite
- Operational capacity
