#**PART 1 — Classification Setup**

**What is Classification?**

- Classification is a supervised learning problem where the target variable belongs to predefined categories.


**Unlike regression:**

- Regression predicts:

  - Continuous values

**Examples:**

- House Price → ₹45,00,000
- Temperature → 32.5°C
- Revenue → ₹10,00,000

**Classification predicts:**

- Categories

**Examples:**
| Input        | Output            |
| ------------ | ----------------- |
| Email        | Spam / Not Spam   |
| Transaction  | Fraud / Normal    |
| Patient Data | Disease / Healthy |
| Customer     | Churn / No Churn  |


**Binary Classification**

- Most introductory classification problems involve:

  - y ∈ {0,1}

**Where:**

- 0 = Negative Class
- 1 = Positive Class

**Examples:**
| Problem            | 0        | 1       |
| ------------------ | -------- | ------- |
| Spam Detection     | Not Spam | Spam    |
| Fraud Detection    | Normal   | Fraud   |
| Disease Prediction | Healthy  | Disease |
| Customer Churn     | Stay     | Leave   |


**Core Objective**

- Instead of predicting:

  - 0 or 1 directly

**we predict:**

- $$P(y=1|x)$$

**Meaning:**

- Probability that the observation belongs to the positive class.

**Example**

- Suppose a model predicts:

  - $$P(Fraud)=0.92$$

**Interpretation:**

- 92% probability that transaction is fraudulent.

**Why Predict Probabilities Instead of Classes?**

- This is one of the most common interview questions.

**Hard Classification**
Fraud

- Only gives:

  - Yes / No

**Probabilistic Classification**
- Fraud Probability = 92%

   - Provides:

    - confidence
    - uncertainty
    - risk estimation


**Advantages**

**1. Captures Uncertainty**

- Prediction:

  - 0.51

- is very different from:

  - 0.99

- Both may belong to class 1.

- But confidence differs dramatically.

**2. Threshold Tuning**

- Business teams can decide:

   - Threshold = 0.5

or

   - Threshold = 0.9

depending on requirements.

**3. Better Decision Making**

- Probability allows:

  - risk analysis
  - ranking
  - prioritization

#**PART 2 — Why Not Linear Regression?**

**Linear Regression Attempt**

- Suppose we try:




image.png

**Example**

- Student Exam Scores
| Score | Pass |
| ----- | ---- |
| 30    | 0    |
| 45    | 0    |
| 55    | 1    |
| 80    | 1    |


**Linear regression attempts:**

- score → pass probability

using a straight line.

**Problem 1: Invalid Probabilities**

- Predictions might become:

  - -0.4

or

  - 1.8

Both are impossible probabilities.



**Problem 2: No Probabilistic Meaning**

- Probability must satisfy:

  - 0 ≤ P ≤ 1

Linear regression cannot guarantee this.

**Problem 3: Sensitive to Outliers**

- A few extreme values can significantly distort predictions.


**Key Insight**

- We need a function that converts:

  - (-∞ , +∞)

to

  - (0 , 1)

**This function is:**

- Sigmoid Function

#**PART 3 — Logistic (Sigmoid) Function**

**Logistic Function**

- The heart of Logistic Regression is:

image.png

**Where:**

- $$z = β₀ + β₁x₁ + β₂x₂ + ... + βₙxₙ$$

**Sigmoid Visualization**

S-shaped curve.

**Python Visualization Code**

```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-10, 10, 100)

sigmoid = 1 / (1 + np.exp(-z))

plt.plot(z, sigmoid)
plt.xlabel("z")
plt.ylabel("Probability")
plt.title("Sigmoid Function")
plt.grid(True)
plt.show()
```

**Properties of Sigmoid**

**Property 1**

- Output always lies in:

  - (0,1)

Perfect for probabilities.

**Property 2**

- Smooth & Differentiable

Essential for optimization.

**Property 3**

- Monotonic

Higher score ⇒ Higher probability.

**Property 4**

- Interpretable

Can be linked directly to odds.


**Why Sigmoid Specifically?**

- Answer
  - Sigmoid converts linear scores into probabilities while preserving interpretability through log-odds.


**Bonus Insight**

- Derivative is elegant:

$$σ
′
(z)=σ(z)(1−σ(z))$$

This makes optimization computationally efficient.

#**PART 4 — Probability Interpretation**

**Understanding Model Output**

- Logistic Regression predicts:

  - Probability

not class labels.

**Example Predictions**
| Output | Interpretation          |
| ------ | ----------------------- |
| 0.10   | 10% positive            |
| 0.40   | 40% positive            |
| 0.80   | 80% positive            |
| 0.95   | Very confident positive |


**Example**

- Prediction:

  - 0.8

**means:**

- 80% chance of belonging to class 1

**NOT:**

- 80% accuracy

Common beginner mistake.

**Thresholding**

- Convert probability into class.


**Default:**
- Threshold = 0.5

**Rule:**

```python
# Example probability from a model
probability = 0.73

if probability >= 0.5:
    predict = 1
else:
    predict = 0

print(predict)
```

**Business Perspective**

- Threshold depends on:

  - risk
  - cost
  
**business objectives**
- Fraud Detection

  - Use:

    - Threshold = 0.9

to reduce false alarms.

**Cancer Detection**

- Use:

  - Threshold = 0.2

to avoid missing patients.

#**PART 5 — Odds & Log-Odds**

**What are Odds?**

- Odds compare:

  - Success vs Failure

**Formula:**

$$Odds= \frac{P}{1-P}$$
	​


**Example**

- Probability:

  - 0.8

- Odds:

  - 0.8 / 0.2 = 4

**Meaning:**

- 4:1 in favor of success


**What are Log-Odds?**

- Take logarithm:

  - $$LogOdds=log(\frac{P}{1-P})$$


**Why Use Log-Odds?**

- Probabilities:

  - 0 to 1

- Log-odds:

  - -∞ to +∞

Now linear modeling becomes possible.

#**PART 6 — Decision Boundary**

**Definition**

- Decision boundary separates classes.


**Binary Classification Rule**

- Predict:

**Class 1**

- if:

  - Probability > 0.5
  - At Probability = 0.5

- Log-odds become:

  - 0

- Therefore:

  - Decision boundary occurs when:

$$β_0+β_1x_1+⋯+β_n​x_n=0$$


**Interpretation**

- Decision boundary is:

  - Linear

in feature space.

**Example**

- Two Features:

  - Age
  - Income

- Boundary:

  - Age + Income = Constant

A straight line.


**Non-Linear Boundaries**

- Logistic Regression can model nonlinear patterns using:

  - Polynomial Features
  - Feature Engineering
  - Interaction Features
